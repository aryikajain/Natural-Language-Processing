N-Gram

```python
tokens = [
    "I",
    "love",
    "machine",
    "learning"
]

for i in range(len(tokens)-1):
    print(
        (tokens[i], tokens[i+1])
    )
```

```python
from nltk.util import ngrams

tokens = [
    "I",
    "love",
    "machine",
    "learning"
]

print("Unigrams:")
print(list(ngrams(tokens, 1)))

print("\nBigrams:")
print(list(ngrams(tokens, 2)))

print("\nTrigrams:")
print(list(ngrams(tokens, 3)))
```

```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
from nltk.util import ngrams

sentence = "I love machine learning"

# Step 1: Tokenize
tokens = word_tokenize(sentence)

print("Tokens:", tokens)

# Step 2: Generate bigrams
bigrams = list(ngrams(tokens, 2))

print("Bigrams:", bigrams)
```

All n-grams at once

```python
from nltk.tokenize import word_tokenize
from nltk.util import ngrams

sentence = "I love machine learning"

tokens = word_tokenize(sentence)

for n in range(1, 4):
    print(f"{n}-gram:")
    print(list(ngrams(tokens, n)))
```

Without NLTK Tokenizer

```python
sentence = "I love machine learning"

tokens = sentence.split()

print(tokens)
print(list(ngrams(tokens, 2)))
```
