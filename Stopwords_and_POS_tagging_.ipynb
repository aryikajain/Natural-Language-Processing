Stopwords

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
stop_words=set(stopwords.words("english"))
```

```python
text = "The cat is sitting on the chair"

tokens = text.lower().split()

filtered = [
    word

    for word in tokens

    if word not in stop_words
]

print(filtered)
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "This is a sample sentence for NLP."

doc = nlp(text)

filtered = [
    token.text
    for token in doc
    if not token.is_stop
]

print(filtered)
```

```python

nlp = spacy.load("en_core_web_sm")

text = "This is a sample sentence"

doc = nlp(text)

for token in doc:
    print(
        token.text,
        token.is_stop
    )
```

pos tagging

```python
import nltk

nltk.download("averaged_perceptron_tagger_eng")
nltk.download("punkt")
```

```python
nltk.download("punkt_tab")
```

```python
from nltk.tokenize import word_tokenize
from nltk import pos_tag

text = "The beautiful girl sings beautifully"

tokens = word_tokenize(text)

print(pos_tag(tokens))
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

doc = nlp(
    "The beautiful girl sings beautifully"
)

for token in doc:
    print(
        token.text,
        token.pos_,
        token.tag_
    )
```
