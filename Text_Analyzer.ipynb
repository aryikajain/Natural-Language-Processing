```python
text="A paragraph is a self-contained unit of writing that develops a single main idea. It typically consists of a topic sentence, supporting sentences with evidence or details, and a concluding sentence. Paragraphs provide structure to text, making extended prose organized and readable"
```

```python
def count_words(text):
  words=len(text.split())
  print("Number of words:",words)
count_words(text)
```

```python
def count_characters(text):
  characters = len(text)
  print("Number of characters:", characters)
```

```python
import re
def count_sentences(text):
    sentences = re.split(r'[.!?]', text)
    sentences = [sentence.strip() for sentence in sentences if sentence.strip()]
    num_sentences = len(sentences)
    print("Number of sentences:", num_sentences)
```

```python
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt')
nltk.download('punkt_tab') # Add this line to download the missing resource
def tokenize_words(text):
    tokens=word_tokenize(text)
    print("tokens",tokens)

tokenize_words(text)
```

```python
from nltk.tokenize import sent_tokenize
def tokenize_sentences(text):
    tokens=sent_tokenize(text)
    print("tokens",tokens)

tokenize_sentences(text)
```

```python
from collections import Counter
def word_frequency(text):
  words=text.split()
  word_counts=Counter(words)
  print("words count",word_counts)
word_frequency(text)
```

```python
def count_most_common_words(text, n):
    words = text.split()
    word_counts = Counter(words)
    most_common_words = word_counts.most_common(n)
    print("Most common words:", most_common_words)
count_most_common_words(text, 1)
```

```python
def lowercase(text):
  lowercase=text.lower()
  print(lowercase)
lowercase(text)
```

```python
def uppercase(text):
  uppercase=text.upper()
  print(uppercase)
uppercase(text)
```

```python
clean_text = re.sub(r'\d+', '', text)

print(clean_text)
```

```python
from nltk.stem import PorterStemmer
def stem_words(text):
    stemmer = PorterStemmer()
    words = text.split()
    stemmed_words = [stemmer.stem(word) for word in words]
    print(" ".join(stemmed_words))
```

```python
def stem_word(text):
      stemmer = PorterStemmer()
      words = text.split()
      for word in words:
        s_word=stemmer.stem(word)
        print(s_word)
stem_word(text)
```
